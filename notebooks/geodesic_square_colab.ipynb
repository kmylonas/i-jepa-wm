# Single-goal geodesic ranking on an expanding square grid

Run on a Colab GPU. The goal is fixed at **(15, 15)** and the wall layout is **(32, 30)**.

- **Use one fixed grid with 2-unit spacing.** The first square runs from (9, 9) to (21, 21). Every legal intersection, including corners and edges, is a fitting position. The center image is the goal reference.
- **Expand the square without shifting the grid.** The half-widths are **6, 12, 18, 24, 30, 36, 42, and 46**. Each stage adds the legal intersections newly enclosed by the larger square. Solid-wall cells and positions beyond the map are skipped. The last square spans the full map, with fitting positions reaching from coordinate 5 to 59.
- **Keep geodesic labels.** Fit all eligible within-band and across-band pairs with a 0.5-unit shortest-path distance gap. Start from identity and warm-start each later stage. Each stage uses 500 updates.
- **Evaluate on an offset grid.** The held-out grid uses different positions and excludes any rendered-image aliases of fitting points. Full-map heatmaps show legal distance, raw MSE, and the adapted cost after every stage.

```python
from google.colab import drive
from pathlib import Path
from datetime import datetime,timezone
import hashlib,json,shutil,signal,subprocess,sys,time
import torch

drive.mount('/content/drive')
assert torch.cuda.is_available(), 'Select a GPU runtime, then Run all.'
DRIVE=Path('/content/drive/MyDrive')
RESUME_RUN_ID=None
RUN_ID=RESUME_RUN_ID or datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ')
OUTPUT=DRIVE/'datasets/curriculum_cost_adapter/geodesic_square_v1'/RUN_ID
REPO=Path('/content/i-jepa-wm-geodesic-square')
REPO_URL='https://github.com/kmylonas/i-jepa-wm.git'
SOURCE_COMMIT='d258503ae676a3db4cfc20eba1e081574cab6af1'
DINO_ROOT=Path('/content/dino_wm')
DINO_COMMIT='0a9492fa12044b852ae9e001cc74604b79c8bb0c'
ENCODER_DRIVE=DRIVE/'checkpoints/IN1K-vit.h.14-300e.encoder-only.pth.tar'
ARTIFACTS=Path('/content/geodesic-square-artifacts')
ENCODER=ARTIFACTS/'encoder.pth.tar'
ENCODER_SHA='3f37d8a7572e25c2dce9a3072c2fd8cd55ce32dc6edb92a8996b0e4cb140bf98'

def sha256(path):
    h=hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda:stream.read(8*1024*1024),b''):h.update(block)
    return h.hexdigest()

def shell(command,cwd=None):
    command=list(map(str,command));print('$',' '.join(command),flush=True)
    with subprocess.Popen(command,cwd=cwd,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1) as process:
        try:
            for line in process.stdout:print(line,end='',flush=True)
            code=process.wait()
        except BaseException:
            process.send_signal(signal.SIGINT)
            try:process.wait(timeout=10)
            except subprocess.TimeoutExpired:process.terminate()
            raise
    if code:raise subprocess.CalledProcessError(code,command)

print('GPU:',torch.cuda.get_device_name())
print('Run ID:',RUN_ID)
print('Output:',OUTPUT)
```

## Check fitting coverage and held-out ranking

The table lists new and cumulative grid intersections for each band. Correct ranking means the point with the longer legal route has higher latent MSE. The same-extent test compares held-out points on the same square perimeter.

```python
import pandas as pd

stages=report['stages']
display(pd.DataFrame([
    dict(band=s['band'],half_width=s['half_width'],
         new_points=sum(p['band']==s['band']-1 for p in report['design']['training']),
         training_nodes=len(s['training_points']),**s['pair_counts'])
    for s in stages
]).set_index('band'))
rows=pd.DataFrame([
    dict(band=s['band'],cohort=cohort,pairs=results['raw']['pairs'],
         raw_pct=None if results['raw']['correct'] is None else 100*results['raw']['correct'],
         adapted_pct=None if results['adapted']['correct'] is None else 100*results['adapted']['correct'])
    for s in stages for cohort,results in s['metrics'].items()
])
display(rows.round(2))
rows.to_csv(OUTPUT/'ranking_metrics.csv',index=False)
```

## View the full map

The legal-distance panel uses simulation units. Each latent MSE panel has its own unclipped color scale. The star marks the goal, open white circles mark fitted positions, dashed outlines mark the current square, and gray denotes the solid wall. The y-axis increases downward as in the simulator image.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import numpy as np

d=report['design']
points=np.asarray(d['points'])
axis=d['axis']
panels=[('Legal path distance',report['truth'],None),('Raw latent MSE',report['raw'],None)]
panels.extend((f"Band {s['band']} · half-width {s['half_width']}",s['adapted'],s) for s in stages)
cols=3
rows=(len(panels)+cols-1)//cols
fig,axes=plt.subplots(rows,cols,figsize=(14,4.5*rows),constrained_layout=True)
for ax in axes.flat[len(panels):]:
    ax.set_visible(False)
for ax,(title,values,stage) in zip(axes.flat,panels):
    field=np.asarray([np.nan if i is None else values[i] for i in d['grid']]).reshape(len(axis),len(axis))
    img=ax.pcolormesh(axis,axis,field,cmap='viridis',shading='nearest',vmin=0)
    ax.scatter(*d['goal'],marker='*',c='black',edgecolor='white',s=135,zorder=4)
    for top,bottom in [(axis[0],26),(34,axis[-1])]:
        ax.add_patch(Rectangle((29,top),6,bottom-top,facecolor='#424242'))
    if stage:
        ax.scatter(*points[stage['training_points']].T,s=10,facecolors='none',edgecolors='white',linewidths=.6)
        half=stage['half_width']
        x,y=d['goal']
        ax.add_patch(Rectangle((x-half,y-half),2*half,2*half,fill=False,color='white',ls='--',lw=.9))
    ax.set(title=title,xlim=(axis[0],axis[-1]),ylim=(axis[-1],axis[0]),
           aspect='equal',xlabel='Ball x',ylabel='Ball y')
    fig.colorbar(img,ax=ax,shrink=.78,label='Simulation units' if title.startswith('Legal') else 'Latent MSE')
fig.savefig(OUTPUT/'full_map_heatmaps.png',dpi=150)
plt.show()
plt.close(fig)
```
